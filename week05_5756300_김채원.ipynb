```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

torch.manual_seed(0)
```

## 문제 1. 사용자 정의 Dataset 작성 (40점)
아래 조건으로 Dataset 클래스를 작성하세요.

- 데이터 개수는 30개, 입력 특성은 4개
- `__init__` `__len__` `__getitem__` 세 메서드를 모두 구현
- 각 메서드가 하는 일을 주석으로 한 줄씩 설명


```python
# 답안
class MyDataset(Dataset):
    # __init__ => 데이터를 준비
    def __init__(self, X, y):
        self.X = X
        self.y = y

    # __len__ => 전체 개수를 알려줌
    def __len__(self):
        return len(self.X)

    # __getitem__ => i번째 데이터 하나를 돌려줌
    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

X = torch.rand(30, 4)
y = torch.rand(30, 1)

ds = MyDataset(X, y)
```

## 문제 2. DataLoader로 batch 출력 (30점)
문제 1의 Dataset을 DataLoader에 넣고 아래를 출력하세요.

- batch 개수
- 각 batch의 입력 shape과 정답 shape
- 마지막 batch의 크기가 다른 이유 (batch size를 7로 두고 확인)


```python
# 답안

ds = MyDataset(X, y)
loader = DataLoader(ds, batch_size=7, shuffle=True)

# 1. batch 개수 출력
print("batch 개수:", len(loader))

# 2. 각 batch의 입력 shape과 정답 shape
for i, (bx, by) in enumerate(loader):
    print(f"batch {i}  X {tuple(bx.shape)}  y {tuple(by.shape)}")

# 3. 마지막 batch의 크기가 다른 이유:
# 전체 데이터 개수 30개를 batch size 7로 나누면 몫이 4, 나머지가 2가 된다.
# 앞의 4개 배치에는 7개씩 들어가고, 마지막 배치에는 남은 2개만 들어가기 때문에 크기가 달라진다.
```

## 문제 3. batch shape 설명 (10점)

`(batch_size, 특성 수)` 의 각 숫자가 무엇을 뜻하는지 적으면 됩니다.

## 문제 4. 학습 루프 작성 (20점)
문제 1의 Dataset으로 아래를 수행하세요.

1. `nn.Linear` 모델 하나를 만든다
2. 다섯 단계(예측 → 손실 → 초기화 → 역전파 → 수정)로 학습 루프를 작성한다
3. epoch마다 평균 손실을 출력해 손실이 내려가는지 확인한다


```python
# 답안
model = nn.Linear(4, 1)
opt = torch.optim.SGD(model.parameters(), lr=0.05)
loss_fn = nn.MSELoss()

for epoch in range(20):
    total = 0.0
    for bx, by in loader:
        pred = model(bx) # 예측
        loss = loss_fn(pred, by) # 손실
        opt.zero_grad() # 초기화
        loss.backward() # 역전파
        opt.step() # 수정

        total += loss.item() * len(bx)

    print(f"epoch {epoch + 1} 평균 손실 {total/len(ds):.4f}")
```
